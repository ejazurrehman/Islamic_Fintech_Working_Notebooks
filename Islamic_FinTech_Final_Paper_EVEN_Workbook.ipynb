### Task 6.3: Analyse Total Financing Over Time

Using Total financing from Indicator Description:

- Filter the dataset for Total Financing.
- Calculate the average Total Financing in USD Millions for each reporting period.
- Display the results in chronological order.

```python
# Filter the dataset for Total Financing
financing_time = df[
    df["Indicator Description"] == "Total financing"
].copy()

# Convert financial value to numeric
financing_time["Values in USD Millions"] = pd.to_numeric(
    financing_time["Values in USD Millions"],
    errors="coerce"
)

# Average financing by reporting period
quarterly_financing = (
    financing_time
    .groupby("Quarter")["Values in USD Millions"]
    .mean()
    .sort_index()
)

display(quarterly_financing)
```

### Task 6.4: Visualize the Financing Trend

```python
# Create a line plot for average financing over time
plt.figure(figsize=(12, 5))

# Plotting the average total financing over time
plt.plot(
    quarterly_financing.index.astype(str),
    quarterly_financing.values,
    marker="o"
)

plt.title("Average Total Financing Over Time")
plt.xlabel("Reporting Period")
plt.ylabel("Average Total Financing (USD Millions)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Task 7.4: Identify Strong Relationships

Using the correlation matrix, identify:

- The pair of numerical variables with the strongest positive correlation.
- The pair of numerical variables with the strongest negative correlation, if one exists.
- One pair with a relatively weak correlation.

```python
# Extract the upper triangle of the correlation matrix
upper = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

# Convert correlations into a long-format table
corr_pairs = (
    upper
    .stack()
    .reset_index()
)

corr_pairs.columns = ["Variable 1", "Variable 2", "Correlation"]

# Sort by correlation
corr_pairs_sorted = corr_pairs.sort_values(
    "Correlation",
    ascending=False
)

display(corr_pairs_sorted.head(10))
```

### Strongest positive correlation

```python
strongest_positive = corr_pairs.loc[
    corr_pairs["Correlation"].idxmax()
]

display(strongest_positive)
```

### Strongest negative correlation

```python
strongest_negative = corr_pairs.loc[
    corr_pairs["Correlation"].idxmin()
]

display(strongest_negative)
```

### Weakest relationship

```python
weakest_relationship = corr_pairs.loc[
    corr_pairs["Correlation"].abs().idxmin()
]

display(weakest_relationship)
```

Create a binary feature called:
